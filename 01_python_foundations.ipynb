## 1. Word Frequency

```python
text = "AI is useful and AI is interesting"
def word_frequency(text):
  words = text.lower().split()
  answer={}
  for word in words:
    answer[word] = answer.get(word, 0) + 1
  return answer
print(word_frequency(text))
print(word_frequency("AI is useful and AI is interesting"))
print(word_frequency("Python Python AI"))
print(word_frequency(""))
```

## 2. Student Filtering

```python
students = [
    {"name": "Gulnur", "score": 85},
    {"name": "Bexultan", "score": 92},
    {"name": "Guzel'", "score": 60},
    {"name": "Rasul", "score": 70},
]
def filter_students(students, minimum_score):
  answer=[student for student in students if student["score"]>=minimum_score]
  return answer
print(filter_students(students, 80))
```

## 3. Duplicate Detection

```python
def find_duplicates(a):
  result = []
  unique = []
  for i in range(len(a)):
    if a[i] in result:
      continue
    else:
      if a[i] in unique:
        result.append(a[i])
      else:
        unique.append(a[i])
  return result
print(find_duplicates([1, 2, 3, 2, 4, 1, 2]))
print(find_duplicates([]))               # []
print(find_duplicates([1, 2, 3]))        # []
print(find_duplicates([5, 5, 5, 5]))      # [5]
print(find_duplicates([3, 2, 2, 3]))     # [2, 3]
```

## 4. Min-Max Normalization

```python
def min_max_normalize(numbers):
  if numbers == []:
    raise ValueError("The list cannot be empty")
  min = numbers[0]
  max = numbers[0]
  new = []
  for i in range(1, len(numbers)):
    if numbers[i]<min:
      min = numbers[i]
    if numbers[i]>max:
      max=numbers[i]
  if min == max:
    raise ValueError("All values are the same")
  for i in range(len(numbers)):
    new.append((numbers[i]-min)/(max-min))
  return new
print(min_max_normalize([10, 20, 30, 40, 50]))
print(min_max_normalize([]))
print(min_max_normalize([5,5,5]))
```

## 5. Train/Test Split

```python
import random
def train_test_split_list(data, test_size=0.2, seed=42):
  if not 0<test_size<1:
    raise ValueError ("test_size should be between 0 and 1")
  rng = random.Random(seed)
  shuffled=data.copy()
  rng.shuffle(shuffled)

  test_count = int(len(data) * test_size)

  if test_count == 0 or test_count == len(data):
    raise ValueError ("data can't be divided like that")

  test = shuffled[:test_count]
  train = shuffled[test_count:]

  return train, test
data = list(range(10))
train, test = train_test_split_list(data)

print("Train:", train)
print("Test:", test)
print("Original:", data)

train, test = train_test_split_list(list(range(10)))
print(len(train), len(test))

```

```python
data = list(range(10))

train1, test1 = train_test_split_list(data, seed=42)
train2, test2 = train_test_split_list(data, seed=42)

assert train1 == train2
assert test1 == test2

assert data == list(range(10))
assert len(train1) == 8
assert len(test1) == 2

assert sorted(train1 + test1) == data

print("All checks passed!")
```
