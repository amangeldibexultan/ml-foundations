```python
scores = np.array([
    [80, 90, 100],
    [60, 70, 80],
    [90, 90, 90]
])

averages = scores.mean(axis=1)

print(scores.mean(axis=1))
print(scores.mean(axis=0))
print(scores.max(axis=1))
print(scores[averages>85])
```

```python
values = np.array([10, 20, 30, 40, 50])
min = values.min()
max = values.max()
normalized = (values-min)/(max-min)
print(normalized)
```

```python
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])
column = data[:,0].copy()
column[:] = 0
print(data)
print(column)
```

```python
features = np.array([
    [20, 1000],
    [30, 2000],
    [40, 3000],
    [50, 4000]
])
new = features[features[:,0]>=30].copy()
min_value = new.min(axis=0)
max_value = new.max(axis = 0)

normalized = (new-min_value)/(max_value-min_value)

print(normalized)
```

```python
data = np.array([
    [20, 1000],
    [30, 1000],
    [40, 1000]
])

def normalize_columns(data):
  min_value = data.min(axis = 0)
  max_value = data.max(axis = 0)
  ranges = max_value - min_value
  ranges[ranges==0] = 1
  return (data-min_value)/(ranges)

print(normalize_columns(data))

expected = np.array([
    [0.0, 0.0],
    [0.5, 0.0],
    [1.0, 0.0]
])

result = normalize_columns(data)

assert np.allclose(result, expected)
print("Normalization check passed!")
```
