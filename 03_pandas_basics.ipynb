```python
import pandas as pd

students = pd.DataFrame({
    "name": ["Cristiano", "Leonel", "Gareth", "Robert"],
    "score": [85, 92, 60, 70],
    "hours_studied": [6, 8, 3, 4]
})

print(students)
```

```python
print(students["name"])
print(students[["name","hours_studied"]])
print(students["score"].mean())
```

```python
print(students[students["score"]>=70])
print(students[
    (students["hours_studied"]>4)&(students["score"]>=85)
])
```

```python
students["score_per_hour"] = students["score"]/students["hours_studied"]
students["passed"] = students["score"] >= 70
print(students)
```

```python
print(students.sort_values("hours_studied", ascending=True))
```

```python
students["group"] = ["A", "A", "B", "B"]
print(students.groupby("group")["score"].mean())
print(students.groupby("group")["hours_studied"].mean())
```

```python
data = pd.DataFrame({
    "name": ["Cristiano", "Leonel", "Gareth", "Robert"],
    "score": [85, 92, None, 70]
})

print(data)
print(data.isna().sum())


data["score"] = data["score"].fillna(data["score"].mean())
print(data)
print(data.isna().sum())
```

```python
data = pd.DataFrame({
    "name": ["Cristiano", "Leonel", "Cristiano", "Robert"],
    "score": [85, 92, 85, 70]
})

print(data.duplicated())
clean_data = data.drop_duplicates()
print(clean_data)
```

```python
data = pd.DataFrame({
    "name": ["Cristiano", "Leonel", "Cristiano", "Gareth", "Robert"],
    "score": [85, 92, 85, 70, None]
})
clean_data = data.drop_duplicates().copy()
clean_data["score"] = clean_data["score"].fillna(clean_data["score"].mean())
clean_data = clean_data.reset_index(drop = True)
print(clean_data)
```

```python
clean_data.to_csv("students_clean.csv", index = False)
```

```python
loaded_data = pd.read_csv("students_clean.csv")
print(loaded_data)
print(loaded_data.shape)
```

```python
def clean_students (data):
  clean_data = data.drop_duplicates().copy()
  clean_data["score"] = clean_data["score"].fillna(clean_data["score"].mean())
  clean_data = clean_data.reset_index(drop = True)
  return clean_data
```

```python
result = clean_students(data)

print(result)

assert len(result) == 4
assert result["score"].isna().sum() == 0
assert result.duplicated().sum() == 0

print("All checks passed!")
expected_mean = (85 + 92 + 70) / 3
actual_score = result.loc[result["name"] == "Robert", "score"].iloc[0]

assert abs(actual_score - expected_mean) < 1e-9
```
