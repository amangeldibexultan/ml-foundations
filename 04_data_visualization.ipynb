# Data Visualization

Practice creating and interpreting plots with Matplotlib.

## Topics
- Bar charts for comparing categories
- Scatter plots with player annotations
- Histograms with custom bin boundaries
- Line plots for results across games
- Comparing two players with a legend
- Saving plots as PNG files

Examples use practice data and are not verified sports statistics.

```python
import pandas as pd
import matplotlib.pyplot as plt

players = pd.DataFrame({
    "name": ["Lebron", "Curry", "Durant", "Edwards"],
    "score": [61, 62, 55, 55],
    "games_played": [1625, 1072,1205, 446]
})
```

```python
plt.bar(players["name"], players["score"])
plt.xlabel("Name")
plt.ylabel("Score")
plt.title("Highest point through the career")
plt.show()
```

```python
plt.bar(players["name"], players["games_played"])
plt.xlabel("Name")
plt.ylabel("Games Played")
plt.title("Players Number of Games")
plt.show()
```

```python
plt.scatter(players["games_played"], players["score"])
plt.xlabel("Games Played")
plt.ylabel("Career-High Points in a Single Game")
plt.title("Games Played and Scores")
plt.annotate("Curry", (1072, 62), xytext = (5,5), textcoords="offset points")
plt.annotate("Lebron", (1625, 61), xytext = (5,5), textcoords="offset points")
plt.annotate("Durant", (1205, 55), xytext = (5,5), textcoords="offset points")
plt.annotate("Edwards", (446, 55), xytext = (5,5), textcoords="offset points")
plt.show()
```

```python
for name, games, score in zip(
    players["name"],
    players["games_played"],
    players["score"]
):
  plt.annotate(name, (games,score), xytext = (5,5), textcoords="offset points")
plt.scatter(players["games_played"], players["score"])
plt.xlabel("Games Played")
plt.ylabel("Career-High Points in a Single Game")
plt.title("Games and Scores")
plt.show()
```

```python
points = [18, 22, 25, 19, 30, 27, 22, 35, 28, 24]

plt.hist(points, bins=5, edgecolor="black")
plt.xlabel("Points per Game")
plt.ylabel("Number of Games")
plt.title("Distribution of Points")
plt.show()
```

```python
plt.hist(points, bins=[15, 25, 35, 45], edgecolor="black")
plt.xticks([15, 25, 35, 45])
plt.xlabel("Points per Game")
plt.ylabel("Number of Games")
plt.title("Distribution of Points")
plt.show()
```

```python
points = [12, 18, 20, 24, 25, 29, 30, 35, 39, 40]
plt.hist(points, bins=[10,25,40], edgecolor = "black")
plt.xticks([10,25,40])
plt.xlabel("Points per Game")
plt.ylabel("Games Played")
plt.title("Distribution of Points")
plt.show()
```

```python
games = [1, 2, 3, 4, 5]
points = [18, 25, 20, 30, 27]

plt.plot(games, points, marker="o")
plt.xticks([1,2,3,4,5])
plt.xlabel("Game Number")
plt.ylabel("Points")
plt.title("Points Across Games")
plt.show()
```

```python
games = [1, 2, 3, 4, 5, 6]
points = [15, 22, 18, 28, 24, 32]
plt.plot(games, points, marker = "o")
plt.xticks([1,2,3,4,5,6])
plt.xlabel("Game Number")
plt.ylabel("Points")
plt.show()
```

```python
points_a = [15, 22, 18, 28, 24, 32]
points_b = [20, 18, 25, 23, 30, 27]
plt.plot([1,2,3,4,5,6], points_a, marker = "o", label="Player A")
plt.plot([1,2,3,4,5,6], points_b, marker = "o", label="Player B")
plt.xticks([1,2,3,4,5,6])
plt.xlabel("Game Number")
plt.ylabel("Points")
plt.title("Comparing Player A and Player B")
plt.legend()
plt.tight_layout()
plt.savefig("player_comparasion.png", dpi=150)
plt.show()
```
